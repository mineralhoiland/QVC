# B2  Lindblad scan on the truncated GPE

Number damping and local dephasing are added **after** each unitary split-step of the truncated 2D GPE (kinetic + cubic-quintic + $V_{\mathrm{ZPF}}$ + $V_{\mathrm{ph}}$). Chern–Simons, Berry, and Skyrme stay **out** of the EOM.

Scan $\tau_{\mathrm{dec}}=5,20,50\,\mathrm{ps}$ with $\gamma_n=\gamma_\varphi=1/\tau_{\mathrm{dec}}$. Report $\Delta_{\mathrm{live}}(1\,\mathrm{ps})$ and $g^{(1)}(1\,\mathrm{ps})$ versus the coherent baseline $\Delta_{\mathrm{live}}\approx 0.6305\,\mathrm{meV}$, $g^{(1)}\approx 0.8832$ (coupled 128² GPE+DCE). A 20 ps window (frozen $V_{\mathrm{ZPF}}$, no DCE) is included because $1\,\mathrm{ps}\ll\tau_{\mathrm{dec}}$.

$\Delta_{\mathrm{live}}$ is computed from $\psi$. There is **no** prescribed $\Delta(t)$ and **no** restored 92% / $0.032$ / $0.389$.

```python
import os, sys
os.environ.setdefault("MPLBACKEND", "Agg")
os.environ.setdefault("MPLCONFIGDIR", "/tmp/mpl-qvc-l1")
from pathlib import Path
from IPython.display import display, Image, Markdown

ROOT = Path.cwd().resolve()
if ROOT.name == "gij_lindblad":
    ROOT = ROOT.parents[1]
elif (ROOT / "notebooks" / "gij_lindblad").is_dir():
    pass
else:
    ROOT = Path("/Users/mineralhoiland/Code/QVCCursor")
sys.path.insert(0, str(ROOT))

from qvc_l1.bootstrap import ensure_qvccompute
ensure_qvccompute()
from qvc_l1.locks import build_l1_locks, COHERENT_DELTA_LIVE_1PS_MEV, COHERENT_G1_1PS
from qvc_l1.lindblad import run_b2
from qvc_l1.figures import fig_b2_delta, fig_b2_g1

OUT = ROOT / "notebooks" / "gij_lindblad" / "output"
OUT.mkdir(parents=True, exist_ok=True)
locks = build_l1_locks()
print("coherent baseline target: Δ_live(1 ps)=", COHERENT_DELTA_LIVE_1PS_MEV, "meV,  g1=", COHERENT_G1_1PS)
print("truncated GPE; CS/Berry/Skyrme off")
```

```python
b2 = run_b2(locks)  # 1 ps at 128² + DCE; 20 ps at 96² frozen V_ZPF
print("this-run coherent 1 ps:", b2["this_run_coherent_1ps"])
print("\n1 ps table")
for row in b2["table_1ps"] + b2["channel_split_1ps"]:
    print(f"  {row['label']:22s} {row['channel']:18s}  Δ={row['Delta_live_meV']}	 g1={row['g1']}")
print("\n20 ps table")
for row in b2["table_20ps"]:
    print(f"  {row['label']:22s} {row['channel']:18s}  Δ={row['Delta_live_meV']}	 g1={row['g1']}")
print("\ncomputed vs ansatz:", b2["computed_vs_ansatz"])
```
